```python
# --- 0) imports & config ---
import pandas as pd
import numpy as np
from datetime import datetime

# tuneables
BASE_RATING = 1500.0
K_BASE = 32.0                 # try 24–48
USE_TIME_DECAY = True
GAMMA = 0.985                 # monthly decay on K (0.98–0.995); only used if USE_TIME_DECAY=True

# --- 1) load data ---
# adjust to your file & column names
df = pd.read_csv("../data/clean_fights.csv")

# expected columns; rename here if yours differ
# examples you might have in your dataset:
# 'f_1_fighter', 'f_2_fighter', 'winner', 'date'
COL_RED  = "f_1_name"
COL_BLUE = "f_2_name"
COL_WIN  = "winner"           # values like the red fighter's name or 'Red'/'Blue'
COL_DATE = "event_date"

# if winner is textual fighter name, convert to 'Red'/'Blue'
def winner_to_side(row):
    if row[COL_WIN] == row[COL_RED] or str(row[COL_WIN]).strip().lower() == "red":
        return "Red"
    elif row[COL_WIN] == row[COL_BLUE] or str(row[COL_WIN]).strip().lower() == "blue":
        return "Blue"
    elif str(row[COL_WIN]).lower() in ["draw", "draws"]:
        return "Draw"
    else:
        return "Unknown"

df["win_side"] = df.apply(winner_to_side, axis=1)

# parse date & sort
df[COL_DATE] = pd.to_datetime(df[COL_DATE], errors="coerce")
df = df.dropna(subset=[COL_DATE]).sort_values(COL_DATE).reset_index(drop=True)

# --- 2) helper functions ---
def expected_score(r_a, r_b):
    """Logistic expected score for A vs B (base-10 Elo)."""
    return 1.0 / (1.0 + 10 ** (-(r_a - r_b) / 400.0))

def k_for_row(prev_date, curr_date, k_base=K_BASE, gamma=GAMMA):
    """Optional time-decayed K based on months elapsed."""
    if not USE_TIME_DECAY or prev_date is None:
        return k_base
    delta_months = max(0, (curr_date.year - prev_date.year) * 12 + (curr_date.month - prev_date.month))
    return k_base * (gamma ** delta_months)

# --- 3) iterate fights & update ratings ---
ratings = {}              # fighter -> rating
last_fight_date = {}      # fighter -> last date they fought (for optional per-fighter decay of K)

# we will also record predictions to evaluate
pred_probs = []
actuals = []

for _, row in df.iterrows():
    red = str(row[COL_RED]).strip()
    blue = str(row[COL_BLUE]).strip()
    d = row[COL_DATE]
    outcome = row["win_side"]   # 'Red', 'Blue', 'Draw', 'Unknown'

    # initialize fighters if new
    r_red = ratings.get(red, BASE_RATING)
    r_blue = ratings.get(blue, BASE_RATING)

    # expected win prob for red
    e_red = expected_score(r_red, r_blue)
    e_blue = 1.0 - e_red

    # store prediction & actual
    pred_probs.append(e_red)
    if outcome == "Red":
        actuals.append(1.0)
        s_red = 1.0; s_blue = 0.0
    elif outcome == "Blue":
        actuals.append(0.0)
        s_red = 0.0; s_blue = 1.0
    elif outcome == "Draw":
        actuals.append(0.5)
        s_red = 0.5; s_blue = 0.5
    else:
        # skip fights with unknown outcome
        continue

    # choose K (time-decayed using the more recent of the two fighters' last-fight dates)
    k_red  = k_for_row(last_fight_date.get(red),  d)
    k_blue = k_for_row(last_fight_date.get(blue), d)

    # updates
    r_red_new  = r_red  + k_red  * (s_red  - e_red)
    r_blue_new = r_blue + k_blue * (s_blue - e_blue)

    ratings[red]  = r_red_new
    ratings[blue] = r_blue_new
    last_fight_date[red]  = d
    last_fight_date[blue] = d

# --- 4) evaluate baseline ---
pred = np.array(pred_probs)
y    = np.array(actuals)

# accuracy if we threshold at 0.5 and ignore draws in denominator
mask_bin = (y == 0) | (y == 1)
acc = ( ((pred >= 0.5) & (y == 1)) | ((pred < 0.5) & (y == 0)) )[mask_bin].mean()

# Brier score (proper scoring rule; lower is better)
brier = np.mean((pred - y)**2)

print(f"Baseline Elo — Accuracy (binary fights only): {acc:.3f}")
print(f"Brier score (all incl. draws=0.5): {brier:.3f}")

# --- 5) build a ratings dataframe & save ---
elo_df = pd.DataFrame([
    {"fighter": f, "elo": r, "last_fight": last_fight_date.get(f)}
] for f, r in ratings.items()).sort_values("elo", ascending=False)

elo_df.head(10)
elo_df.to_json("../models/elo_ratings.json", orient="records")
print("Saved fighter ratings → ../models/elo_ratings.json")
```

```python
# --- 0) imports & config ---
import pandas as pd
import numpy as np

# tuneables
BASE_RATING = 1500.0
K_BASE = 32.0                 # try 24–48
USE_TIME_DECAY = True
GAMMA = 0.985                 # monthly decay on K (0.98–0.995); only used if USE_TIME_DECAY=True

# --- 1) load data ---
df = pd.read_csv("../data/UFC_full_data_golden.csv")

# expected columns in YOUR file
COL_RED  = "f_1_name"
COL_BLUE = "f_2_name"
COL_WIN  = "winner"
COL_DATE = "event_date"       # change if your date col differs

# --- winner mapping: map to 'Red' / 'Blue' / 'Draw' / 'Unknown' ---
def winner_to_side(row):
    w = row.get(COL_WIN, None)
    r = str(row.get(COL_RED, "")).strip().lower()
    b = str(row.get(COL_BLUE, "")).strip().lower()

    if pd.isna(w):
        return "Unknown"
    w_str = str(w).strip().lower()

    # direct name matches or explicit corner labels
    if w_str == r or w_str == "red":
        return "Red"
    if w_str == b or w_str == "blue":
        return "Blue"

    # draw variants
    if "draw" in w_str:
        return "Draw"

    return "Unknown"

df["win_side"] = df.apply(winner_to_side, axis=1)

# --- parse date & sort chronologically ---
df[COL_DATE] = pd.to_datetime(df[COL_DATE], errors="coerce")
df = df.dropna(subset=[COL_DATE]).sort_values(COL_DATE).reset_index(drop=True)

# --- 2) helper functions ---
def expected_score(r_a, r_b):
    """Logistic expected score for A vs B (base-10 Elo)."""
    return 1.0 / (1.0 + 10 ** (-(r_a - r_b) / 400.0))

def k_for_row(prev_date, curr_date, k_base=K_BASE, gamma=GAMMA):
    """Optional time-decayed K based on months elapsed."""
    if not USE_TIME_DECAY or prev_date is None:
        return k_base
    delta_months = max(0, (curr_date.year - prev_date.year) * 12 + (curr_date.month - prev_date.month))
    return k_base * (gamma ** delta_months)

# --- 3) iterate fights & update ratings ---
ratings = {}              # fighter -> rating
last_fight_date = {}      # fighter -> last fight date

pred_probs = []           # predictions we made
actuals = []              # corresponding labels (1, 0, or 0.5)

for _, row in df.iterrows():
    red  = str(row[COL_RED]).strip()
    blue = str(row[COL_BLUE]).strip()
    d    = row[COL_DATE]
    outcome = row["win_side"]   # 'Red', 'Blue', 'Draw', 'Unknown'

    # skip if we don't have a clean outcome
    if outcome == "Unknown":
        continue

    # init ratings if new
    r_red  = ratings.get(red,  BASE_RATING)
    r_blue = ratings.get(blue, BASE_RATING)

    # expected probs (before update)
    e_red  = expected_score(r_red, r_blue)

    # append prediction & label ONLY when outcome is known
    pred_probs.append(e_red)
    if outcome == "Red":
        s_red, s_blue = 1.0, 0.0
        actuals.append(1.0)
    elif outcome == "Blue":
        s_red, s_blue = 0.0, 1.0
        actuals.append(0.0)
    else:  # Draw
        s_red, s_blue = 0.5, 0.5
        actuals.append(0.5)

    # choose per-fighter K (optionally time-decayed)
    k_red  = k_for_row(last_fight_date.get(red),  d)
    k_blue = k_for_row(last_fight_date.get(blue), d)

    # updates
    ratings[red]  = r_red  + k_red  * (s_red  - e_red)
    ratings[blue] = r_blue + k_blue * (s_blue - (1.0 - e_red))
    last_fight_date[red]  = d
    last_fight_date[blue] = d

# --- 4) evaluate baseline (safe lengths, ignore draws in accuracy) ---
n = min(len(pred_probs), len(actuals))
pred = np.array(pred_probs[:n])
y    = np.array(actuals[:n])

mask_bin = (y == 0) | (y == 1)
if mask_bin.any():
    acc = ( ((pred >= 0.5) & (y == 1)) | ((pred < 0.5) & (y == 0)) )[mask_bin].mean()
else:
    acc = float("nan")

brier = np.mean((pred - y)**2)

print(f"Used {n} fights | Binary for accuracy: {mask_bin.sum()}")
print(f"Baseline Elo — Accuracy (binary only): {acc:.3f}")
print(f"Brier score (incl. draws at 0.5): {brier:.3f}")

# --- 5) build a ratings dataframe & save ---
elo_df = pd.DataFrame(
    [{"fighter": f, "elo": r, "last_fight": last_fight_date.get(f)} for f, r in ratings.items()]
).sort_values("elo", ascending=False)

display(elo_df.head(10))
elo_df.to_json("../models/elo_ratings.json", orient="records")
print("Saved fighter ratings → ../models/elo_ratings.json")
```

```python
import json
import math

BASE_RATING = 1500.0

def expected_prob(r_red, r_blue):
    return 1.0 / (1.0 + 10 ** (-(r_red - r_blue)/400.0))

def load_ratings(path="../models/elo_ratings.json"):
    with open(path, "r") as f:
        items = json.load(f)
    return {row["fighter"]: float(row["elo"]) for row in items}

def predict_prob(red_name, blue_name, ratings):
    r_red  = ratings.get(red_name, BASE_RATING)
    r_blue = ratings.get(blue_name, BASE_RATING)
    return expected_prob(r_red, r_blue)

# load once per session
RATINGS = load_ratings()

# example calls
print("P(Jon Jones beats Islam Makhachev) =", round(predict_prob("Jon Jones", "Islam Makhachev", RATINGS), 3))
print("P(Max Holloway beats Kamaru Usman) =", round(predict_prob("Max Holloway", "Kamaru Usman", RATINGS), 3))
```
